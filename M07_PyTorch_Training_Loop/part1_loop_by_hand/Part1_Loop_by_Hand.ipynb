# Part 1: The training loop by hand

Fit a straight line, `y = w * x + b`, to noisy points. There is no library here beyond NumPy, so every step of the loop is visible. The steps are the same in every framework: predict, measure the error, work out which way to move each number, and move.

```python
import numpy as np

rng = np.random.default_rng(0)
x = rng.uniform(0, 5, 50)
y = 3 * x + 2 + rng.normal(0, 0.3, 50)   # the truth is w = 3 and b = 2, plus noise
print("first three points:", np.round(x[:3], 2), np.round(y[:3], 2))
```

## The loop

The loss is the mean squared error. The two gradient lines are what calculus gives for this loss, derived by hand. Part 3 shows how PyTorch finds them for you.

```python
w, b = 0.0, 0.0
learning_rate = 0.05

for step in range(1, 201):
    prediction = w * x + b                      # 1. forward: make predictions
    error = prediction - y
    loss = np.mean(error ** 2)                  # 2. loss: one number for how wrong we are
    grad_w = 2 * np.mean(error * x)             # 3. gradient: which way to move w
    grad_b = 2 * np.mean(error)                 #    and which way to move b
    w -= learning_rate * grad_w                 # 4. update: a small step downhill
    b -= learning_rate * grad_b
    if step in (1, 2, 3, 10, 50, 100, 200):
        print(f"step {step:3d}  loss {loss:8.4f}  w {w:.3f}  b {b:.3f}")
```
