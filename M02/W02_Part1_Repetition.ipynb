```python
# ---------------------------------------------------------------------------
# Simulated input
# ---------------------------------------------------------------------------
# Notebooks cannot answer input() prompts during "Run All": execution simply
# waits for you. So that every cell runs top to bottom without stopping, this
# helper replays a list of pretend answers instead.
#
# To type the answers yourself, set USE_REAL_INPUT = True and re-run this cell.
# In a normal .py script you would just call input() directly.

USE_REAL_INPUT = False


def make_asker(answers):
    """Return an ask() function that replays `answers` in order."""
    replay = iter(answers)

    def ask(prompt):
        if USE_REAL_INPUT:
            return input(prompt)
        answer = next(replay)
        print(f"{prompt}{answer}")      # echo it, so the output reads like a real session
        return answer

    return ask

print("Simulated input ready. USE_REAL_INPUT =", USE_REAL_INPUT)
```

## 1. The `for` loop walks through items

A Python `for` loop does not count. It hands you each item of a collection in turn.

```python
for name in ["Ada", "Grace", "Alan"]:
    print(f"Hello, {name}")
```

## 2. Counting with `range`

When you do want numbers, iterate over a `range`. The **stop value is never included**, which is consistent everywhere in Python, including slicing.

```python
# list() makes the range visible; a range on its own produces values lazily.
print("range(5)        ->", list(range(5)))          # 0 to 4, stop excluded
print("range(2, 10, 3) ->", list(range(2, 10, 3)))   # start, stop, step
print("range(5, 0, -1) ->", list(range(5, 0, -1)))   # a negative step counts down
```

## 3. Position and item together: `enumerate` and `zip`

`for i in range(len(items))` works, but it is the pattern of someone translating from another language. `enumerate` gives you the position; `zip` walks two sequences side by side.

```python
names = ["Ada", "Grace", "Alan"]
scores = [91, 85, 78]

# start=1 numbers from 1, which is what humans expect in a printed list.
for position, name in enumerate(names, start=1):
    print(f"{position}. {name}")

print()

for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

### Watch out: `zip` stops silently at the shorter sequence

No error, no warning. The unmatched item is simply dropped. Run the cell and look for Alan.

```python
names = ["Ada", "Grace", "Alan"]
scores = [91, 85]                      # one score missing

print("Silent truncation:", list(zip(names, scores)))    # Alan has vanished

# strict=True (Python 3.10+) turns the silent problem into a loud one.
# The try/except is here only so this notebook keeps running; in real code
# you would usually let the error stop the program.
try:
    list(zip(names, scores, strict=True))
except ValueError as error:
    print("With strict=True:  ValueError ->", error)
```

## 4. The `while` loop: retry until valid

A `while` loop re-checks its condition before every pass. The classic use is input that must be retried until it is valid, an improvement on Lab 01, where a bad entry ended the program.

The simulated answers below include three bad entries before a good one, so you can watch the loop reject each.

```python
ask = make_asker(["abc", "", "150", "72"])   # letters, empty, out of range, valid

while True:
    raw = ask("Enter a mark out of 100: ")

    # isdigit() is False for "", "abc", "-5" and "7.5", so it doubles as
    # a check that int() will succeed. The range test only runs if it does,
    # because `and` short-circuits.
    if raw.isdigit() and 0 <= int(raw) <= 100:
        mark = int(raw)
        break                              # success: leave the loop

    print("  Please enter a whole number from 0 to 100.")

print(f"Accepted: {mark}")
```

### `continue` skips to the next pass

`break` leaves the loop entirely. `continue` abandons only the current pass.

```python
for n in range(1, 11):
    if n % 3 == 0:
        continue                           # skip multiples of 3
    print(n, end=" ")
print()
```

## 5. Nested loops

The inner loop runs to completion for every pass of the outer one, so the total work is the product of the two. Fine for a 3 × 3 table; a real problem at 10,000 × 10,000.

```python
for row in range(1, 4):
    for col in range(1, 4):
        print(row * col, end="\t")        # \t is a tab, keeping columns aligned
    print()                                # new line after each row
```

### `break` leaves only the loop it is directly inside

A common surprise. Predict the output before running.

```python
for row in range(3):
    for col in range(3):
        if col == 1:
            break                          # leaves the INNER loop only
        print(f"row {row}, col {col}")
    print(f"  outer loop carries on after row {row}")
```
